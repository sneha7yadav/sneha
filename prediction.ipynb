```python
from sklearn.linear_model import LinearRegression
#house size
X=[[1000],[1200],[1500],[1800]]
#house price (in lakh)
y=[40,50,65,80]
model=LinearRegression()
model.fit(X,y)
prediction=model.predict([[1600]])
print("prediction price",prediction[0],"lakh")
```

```python
from sklearn.cluster import KMeans
import numpy as np
data=np.array([[2,3],[3,4],[10,12],[11,13]])
model=KMeans(n_clusters=2,random_state=42)
model.fit(data)
print("Cluster Labels:")
print(model.labels_)
```

```python
from sklearn.cluster import KMeans
import numpy as np
data=np.array([[2,3],[3,4],[10,12],[11,13]])
model=KMeans(n_clusters=1,random_state=42)
model.fit(data)
print("Cluster Labels:")
print(model.labels_)
```

```python
from sklearn.linear_model import LinearRegression
x=([1],[2],[3],[4],[5],[6],[7])
y=[12.5,8.2,15.0,10.4,7.0,9.3,6.8]
model=LinearRegression()
model.fit(x,y)
prediction=model.predict([[8]])
print("Predicted Rainfall for Day 8:", round(prediction[0], 2), "mm")
```

```python
from sklearn.tree import DecisionTreeClassifier

X = [
    [2],
    [3],
    [5],
    [6],
    [7]
]

y = ["Fail", "Fail", "Pass", "Pass", "Pass"]

model = DecisionTreeClassifier()

model.fit(X, y)

prediction = model.predict([[4]])

print(prediction)
```

```python
from sklearn.ensemble import RandomForestClassifier

X = [
    [2],
    [3],
    [5],
    [6],
    [7]
]

y = ["Fail", "Fail", "Pass", "Pass", "Pass"]

model = RandomForestClassifier(n_estimators=100)

model.fit(X, y)

prediction = model.predict([[4]])

print(prediction)
```
